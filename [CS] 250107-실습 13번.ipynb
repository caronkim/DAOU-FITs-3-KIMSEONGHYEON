## 배열의 요소 추가/삽입

```python
arr = [1, 2, 3]

arr.append(4)
print(arr)

arr.insert(1,10)
print(arr)

arr_ones = [1] * 5
print(arr_ones)
```

## 배열 요소 삭제

```python
arr = [10, 20, 30, 40, 50]

arr.remove(30)
print(arr)

arr.pop()
print(arr)

arr.pop(0)
print(arr)
```

## 배열 요소 수정

```python
arr = [1, 2, 3, 4, 5]
arr[2] = 99
arr
```

## 배열 합치기

```python
arr1 = [1, 2, 3]
arr2 = [4, 5, 6]

arr3 = arr1 + arr2
print(arr3)

arr1.extend(arr2)
print(arr1)
```
